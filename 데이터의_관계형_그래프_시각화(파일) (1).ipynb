```python
#노드에 툴팁(마우스로 클릭했을 때 나오는 글자 )추가하기

from pyvis.network import Network
from IPython.display import HTML

#정점 정보 추가
## label과 title에 임의의 값 집어넣고 결과 확인하기
g = Network(notebook=True, cdn_resources='in_line')
g.add_node(0, label='오타니', title='다저스')
g.add_node(1, label='이정후', title='자이언츠')
g.add_node(2, label='류현진', title='한화')

#간선 정보 추가
g.add_edge(0, 1, arrows='to',title="투수->타자",label="투타")
g.add_edge(0, 2, arrows='to',title="투수->투수",label="투투")

#그래프의 정점과 간선 정보 출력
for n in g.nodes:
  print(n)
for n in g.edges:
  print(n)

g.save_graph("g3.html")
HTML(filename="g3.html")
```
